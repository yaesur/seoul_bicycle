```python
import glob
import pandas as pd

files = sorted(glob.glob("../data/raw/2025_0[4-6]/*.csv"))

df_list = []

for f in files:
    df = pd.read_csv(f, encoding="cp949")
    df_list.append(df)

bicycle = pd.concat(df_list, ignore_index=True)

bicycle.shape

# 행 개수 21255255 컬럼 개수 10 월별, 일별 다 concat으로 합침
```

```python
bicycle.columns
```

```python
bicycle.info()
```

```python
bicycle["date"] = pd.to_datetime(
    bicycle["기준_날짜"].astype(str), format = "%Y%m%d"
)
```

```python
bicycle[["기준_날짜", "date"]].head()
```

```python
bicycle["month"] = bicycle["date"].dt.month
bicycle["weekday"] = bicycle["date"].dt.weekday
```

```python
bicycle["start_hour"] = bicycle["기준_시간대"] // 100
bicycle["start_minute"] = bicycle["기준_시간대"] % 100
```

```python
cat_cols = [
    "집계_기준", "시작_대여소_ID", "시작_대여소명", "종료_대여소_ID", "종료_대여소명"
]

for c in cat_cols:
    bicycle[c] = bicycle[c].astype("category")

# 2천만건 이상의 대규모 데이터를 다루면서 분석 과부하를 방지하기 위해 문자열 컬럼을 category 타입으로 변환함
# 메모리 사용량과 처리 속도를 최적화함
```

```python
bicycle.head()
```

```python
weekday_map = {
    0: 'Mon', 1: 'Tue', 2: 'Wed', 3: 'Thu', 4: 'Fri', 5: 'Sat', 6: 'Sun'
}

bicycle['weekday_name'] = bicycle['weekday'].map(weekday_map).astype('category')
```

```python
bicycle['is_commute'] = (
    ((bicycle['start_hour'].between(7,9)) |
     (bicycle['start_hour'].between(18,20)))
     & (bicycle['weekday'] < 5)
)
```

```python
bicycle.head()
```

```python
holidays = [
    '2025-05-05',  # 어린이날 + 석가탄신일
    '2025-05-06',  # 어린이날 대체 공휴일
    '2025-06-03',  # 대통령선거 임시 공휴일
    '2025-06-06'   # 현충일
] 

# 법정 공휴일만 공휴일로 처리 (근로자의 날은 제외)

holidays = pd.to_datetime(holidays) 

# 여기까지가 공휴일만 뽑아 to_datetime

bicycle['is_holiday'] = bicycle['date'].isin(holidays)
bicycle['is_weekend'] = bicycle['weekday'] >= 5 # 토일 라벨링 56을 주말로 지정
bicycle['is_workday'] = (~bicycle['is_weekend']) & (~bicycle['is_holiday'])
```

```python
bicycle.to_csv(
    '../data/processed/bicycle_2025_04_06_cleaned.csv',
    index=False,
    encoding='utf-8-sig'
)

# bicycle 데이터 전처리한 거 상위폴더 data/processed로 옮기기
# 파일명과 폴더명은 전처리로 같은 의미지만 같은 이름을 방지하기 위해
# processed cleaned로 함
```
